## 1. Import the libraries
We use:
- **NumPy** to create and calculate numerical data.
- **Pandas** to store, clean and analyse tabular data.
- **Matplotlib** to create charts.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

## 2. Create a sample dataset
The dataset contains department, number of students, number placed, and average package (in LPA). The values are fictional.

```python
data = {
    "Department": ["CSE", "ECE", "IT", "ME", "CE", "EEE"],
    "Students": [120, 100, 90, 80, 70, 60],
    "Placed": [96, 70, 72, 48, 35, 36],
    "Average_Package_LPA": [6.5, 5.2, 6.0, 4.2, 3.8, 4.5]
}

df = pd.DataFrame(data)
df
```

## 3. Explore and check the data
Use `head()`, `info()` and `describe()` to understand the dataset.

```python
print("First five rows:")
display(df.head())

print("\\nDataset information:")
df.info()

print("\\nNumerical summary:")
display(df.describe())
```

## 4. Calculate placement percentage
Formula:

**Placement Percentage = (Number of students placed / Total students) × 100**

```python
df["Not_Placed"] = df["Students"] - df["Placed"]
df["Placement_Percentage"] = (df["Placed"] / df["Students"]) * 100
























display(df[["Department", "Students", "Placed", "Not_Placed", "Placement_Percentage"]].round(2))
```

## 5. Overall placement summary
NumPy is used here to calculate totals and the overall percentage.

```python
total_students = np.sum(df["Students"].to_numpy())
total_placed = np.sum(df["Placed"].to_numpy())
overall_percentage = (total_placed / total_students) * 100

print("Total students:", total_students)
print("Total students placed:", total_placed)
print("Overall placement percentage:", round(overall_percentage, 2), "%")
```

## 6. Department-wise analysis
Find the department with the highest placement percentage and the highest average package.

```python
best_placement_dept = df.loc[df["Placement_Percentage"].idxmax(), "Department"]
best_package_dept = df.loc[df["Average_Package_LPA"].idxmax(), "Department"]

print("Highest placement percentage department:", best_placement_dept)
print("Highest average package department:", best_package_dept)

display(df.sort_values("Placement_Percentage", ascending=False))
```

## 7. Visualise placement percentage
A bar chart makes department-wise comparison easier.

```python
plt.figure(figsize=(9, 5))
plt.bar(df["Department"], df["Placement_Percentage"])
plt.title("Department-wise Placement Percentage")
plt.xlabel("Department")
plt.ylabel("Placement Percentage (%)")
plt.ylim(0, 100)
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.show()
```

## 8. Compare average package

```python
plt.figure(figsize=(9, 5))
plt.bar(df["Department"], df["Average_Package_LPA"])
plt.title("Average Package by Department")
plt.xlabel("Department")
plt.ylabel("Average Package (LPA)")
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.show()
```

## 9. Export the analysis
This saves the analysed table as a CSV file. In Google Colab, the next cell downloads it.

```python
output_file = "placement_analysis_output.csv"
df.to_csv(output_file, index=False)
print("Saved:", output_file)
```
